```python
import numpy as np
import pandas as pd
import torch
from compactts.agent import CompactTSAgent, SeriesProfiler, OpenMetaRouter

print("CompactTS-Agent modules imported successfully!")
```

## 1. Autonomous Series Profiler
The SeriesProfiler extracts structural properties (autocorrelation, seasonality, trend ratio, kurtosis) in milliseconds.

```python
profiler = SeriesProfiler(freq='H')

# Generate a synthetic time series with trend and seasonality
t = np.linspace(0, 10, 200)
synthetic_series = 2.0 * np.sin(2 * np.pi * t / 24) + 0.5 * t + 0.1 * np.random.randn(200)

profile = profiler.profile(synthetic_series)
for k, v in profile.items():
    print(f"{k:<20s}: {v}")
```

## 2. Meta-Routing Policy
The OpenMetaRouter routes each series to the optimal open foundation expert.

```python
router = OpenMetaRouter()

for domain in ['Energy', 'Web/CloudOps', 'Transport', 'Healthcare', 'Econ/Fin', 'Sales']:
    expert = router.route_domain(domain)
    print(f"Domain: {domain:<15s} -> Routed Expert: {expert}")
```

## 3. Probabilistic Forecasting with CompactTS-Agent

```python
agent = CompactTSAgent(prediction_length=24, freq='H', batch_size=32)
start_dates = [pd.Period('2024-01-01 00:00', freq='h')]

forecasts = agent.forecast_batch([synthetic_series], start_dates)
q_array = forecasts[0].forecast_array

print(f"Forecast generated: {q_array.shape[0]} quantiles across {q_array.shape[1]} horizon steps")
print(f"Median forecast (q0.5):\n{q_array[4]}")
```
